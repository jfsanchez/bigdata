## Conceptos
| Termo | Que é |
| --- | --- |
| **HOME (NFS)** | Disco Linux, `/home/xunta/xueduaXX`, montado en cada nodo de login. Aquí están estes notebooks |
| **HDFS** | Sistema de ficheiros distribuído, `/user/xueduaXX`. De aquí le Hive |
| **YARN** | Reparte CPU e memoria do clúster entre os traballos |
| **Hive** | SQL sobre ficheiros de HDFS; traduce as consultas a traballos MapReduce |
| **Metastore** | Onde Hive garda o esquema e a ruta de cada táboa |
| **beeline** | Cliente que envía o SQL a Hive (HiveServer2). `hive()` úsao por debaixo |
| **Spark** | Outro motor de procesamento; usa o mesmo metastore |
